# Task 1.1: Core Contribution / Architecture of Truncated Gradient

## Step-by-Step Method Description

*   **Step 1:** Initialization
    *   Initialize weights $w_j \leftarrow 0$ for all features $j=1, \dots, d$.
    *   *Reference:* Section 4, Algorithm 1 (Truncated Gradient for Least Squares).
    *   *Purpose:* Begins the online learning process with an empty model (all feature weights are zero).

*   **Step 2:** Example Acquisition
    *   Acquire an unlabeled example $x_i = [x_1, x_2, \dots, x_d]$ from the data stream.
    *   *Reference:* Section 4, Algorithm 1, Step 1.
    *   *Purpose:* Receives the next input point to process in the online (streaming) setting.

*   **Step 3:** Truncation (Gravity Application)
    *   For every feature weight $w_j$:
        *   If $0 < w_j \le \theta$, apply gravity: $w_j \leftarrow \max\{w_j - g_i \eta, 0\}$
        *   If $-\theta \le w_j < 0$, apply gravity: $w_j \leftarrow \min\{w_j + g_i \eta, 0\}$
    *   *Reference:* Section 3.3, Equation 6 and Section 4, Algorithm 1, Step 2.
    *   *Purpose:* This is the *core contribution* of the paper. It actively pulls small weights (those within the range $[-\theta, \theta]$) exactly towards zero at a steady rate ($g_i \eta$). This induces actual vector sparsity, unlike standard L1 regularization in SGD. The parameters control how aggressive this sparsification is.

*   **Step 4:** Prediction
    *   Compute the predicted output $\hat{y} = \sum_{j} w_j x_j$ using the newly truncated weights.
    *   *Reference:* Section 4, Algorithm 1, Step 3.
    *   *Purpose:* Estimates the target value for the current input based on the active, sparse features.

*   **Step 5:** Label Acquisition and Loss
    *   Acquire the true label $y_i$ for the current example from the oracle.
    *   *Reference:* Section 4, Algorithm 1, Step 4.
    *   *Purpose:* Brings in the ground truth to calculate the error of the current model prediction.

*   **Step 6:** Gradient Descent Update
    *   Update all feature weights using standard stochastic gradient descent on the specific loss function (e.g., for square loss: $w_j \leftarrow w_j + 2\eta(y - \hat{y})x_j$).
    *   *Reference:* Section 2, Equation 3; Section 4, Algorithm 1, Step 5.
    *   *Purpose:* Adjusts the weights to minimize the prediction error on the current example.
    
## Final Summary Sentence
This paper solves the problem of online machine learning models consuming excessive memory and test-time computation when dealing with massive datasets containing millions of sparsely relevant features; the authors claim their `truncated gradient` approach is better than existing alternatives (like standard L1 regularization or simple coefficient rounding) because it successfully induces exact sparsity continuously and efficiently while mathematically guaranteeing bounded regret loss compared to standard gradient descent.
