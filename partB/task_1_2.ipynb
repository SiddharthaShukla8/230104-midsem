# Task 1.2: Key Assumptions

### Assumption 1
**Assumption:** The loss function $L(w, z)$ is convex with respect to the weight parameter $w$, and its sub-gradient growth is bounded (i.e., $\|\nabla_1 L(w, z)\|^2 \le A \cdot L(w, z) + B$).

**Why the method needs it:** The theoretical regret bound for the truncated gradient algorithm fundamentally depends on this convexity and bounded gradient growth logic to guarantee that applying the non-convex discontinuous "truncation" step does not cause the online learning process to diverge uncontrollably. Without convexity, the regret bounds (which mirror standard online gradient descent proofs) would fail.

**Violation scenario:** Training a multi-layer deep neural network using the truncated gradient method to sparsify network connections. The loss landscape of deep networks is highly non-convex, violating the fundamental assumption of Theorem 3.1.

**Paper reference:** Section 3.4, Assumption 3.1.

---

### Assumption 2
**Assumption:** The feature space is inherently sparse, meaning that a significant fraction of the $d$ total features in the dataset are irrelevant or uninformative for predicting the target variable.

**Why the method needs it:** The entire objective of the "truncated gradient" algorithm is to pull small weights to exactly zero, sacrificing a small amount of prediction accuracy (controlled by $g$) to gain massive sparsity. If every feature is highly informative (meaning the true optimal weight vector $\bar{w}$ is entirely dense), the truncation mechanism will forcefully drop useful information, strictly lowering accuracy without yielding any meaningful memory or computational savings.

**Violation scenario:** A dataset constructed using Polynomial Regression or synthetic parity functions where the target depends tightly and equally on all 100 features. 

**Paper reference:** Section 1 (Introduction), point exactly about handling "large datasets... with large numbers of features, substantial sparsity is discoverable", and Section 6.1 regarding reducing features on effectively sparse real-world applications (like `rcv1`).

---

### Assumption 3
**Assumption:** The magnitude of the input features $x$ is bounded above by some constant $C$ ($\forall x, \|x\| \le C$).

**Why the method needs it:** For the specific application of truncated gradient to Least Squares regression (Algorithm 1), the paper derives the square loss regret bound (Corollary 4.1). In this bound, the learning rate $\eta$ and the bounds on the loss rely on $C^2$ to ensure that a single gradient update doesn't drastically overshoot the optimal weights when the input magnitudes are huge.

**Violation scenario:** An unbalanced financial dataset where some features represent values in pennies (e.g., `0.02`) and others represent un-normalized market capitalizations in trillions of dollars (e.g., `2,500,000,000,000`). This completely violates the bounded magnitude assumption without prior scaling.

**Paper reference:** Section 3.4 (definition of loss functions parameters) and Section 4, Corollary 4.1 (Sparse Online Square Loss Regret), explicitly stating: "If there exists $C > 0$ such that for all $x, \|x\| \le C$...".
