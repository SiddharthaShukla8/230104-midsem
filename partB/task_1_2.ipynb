# Task 1.2 Key Assumptions (8 marks)

## Assumption 1
Assumption: The loss function $L(w, z)$ is convex in $w$ and there exist non-negative constants $A$ and $B$ such that $\|
abla_1 L(w, z)\|^2 \le A L(w, z) + B$.
Why the method needs it: This mathematically bounds the sum of the magnitudes of the gradients. The performance guarantees (bounding the online regret) of the truncated gradient method critically depend on this property to show that the learning updates converge efficiently toward the optimal regularized weights without diverging.
Violation scenario: Training a complex deep neural network where the loss function with respect to the weights is highly non-convex and full of unpredictable local minima.
Paper reference: Assumption 3.1.

## Assumption 2
Assumption: The feature vectors $x_i$ arriving in the data stream are highly sparse (i.e., the number of non-zero features $k \ll d$).
Why the method needs it: The method's key practical advantage is its efficient implementation (using a "lazy update" trick keeping timestamps $	au_j$). This implementation simulates aggressive sparsity penalty efficiently by only iterating over the non-zero features in each incoming example, requiring $O(k)$ operations.
Violation scenario: Processing raw uncompressed image pixel arrays, where almost all pixels have non-zero intensities, thereby destroying the computational speedup.
Paper reference: Section 5 ("Efficient Implementation").

## Assumption 3
Assumption: The optimal predictive model for the data has a sparse structure, meaning most of the large number of available features are uninformative noise.
Why the method needs it: The primary goal of the method is continuous sparsification (shrinking weights exactly to zero) to save memory and processing time. If all features are strictly required, truncation will inherently penalize necessary information and severely degrade accuracy.
Violation scenario: A highly engineered but low-dimensional dataset (e.g., using 10 well-crafted uncorrelated features to predict a physical property) where removing any feature causes significant loss in model fit. 
Paper reference: Section 1 ("Introduction") and Section 6.1 ("Feature Sparsification of Truncated Gradient").
