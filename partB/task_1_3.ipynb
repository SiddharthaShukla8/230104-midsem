# Task 1.3: What the Paper Claims to Improve

### Baseline Method
The main baseline prior method that the paper compares its approach against is **Stochastic Gradient Descent (SGD) with naive L1-Regularization**, often represented as $f(w_i) = w_i - \eta \nabla_1 L(w_i, z_i) - \eta g \cdot \text{sgn}(w_i)$ (Equation 5 in the text).

### Limitation of the Baseline
The paper identifies that naive L1-regularization applied directly to the online stochastic gradient descent rule completely fails to induce actual sparsity in practice. Because gradient updates consist of continuously adding/subtracting continuous floating-point numbers ($a + b$), it is mathematically highly improbable for a weight parameter to "accidentally" sum exactly to zero. Therefore, standard L1-regularized SGD might produce *small* weights, but it doesn't give *sparse* (exactly zero) weights, rendering it useless for memory reduction on large feature sets.

### Attempted Improvement
The proposed method, Truncated Gradient, overcomes this limitation by actively and artificially shrinking weights that fall within a specified threshold $\theta$ towards zero by a set gravity amount ($g_i \eta$). Crucially, if the weight crosses zero during this pull, it is abruptly truncated (*rounded*) exactly to $0$. This mechanics guarantees exact sparsity.

### Failure Scenario
The truncated gradient method would likely **not outperform** the standard baseline (L1-regularized SGD or even standard unregularized SGD) on a completely dense dataset where every single feature contains independent, crucial signal with absolutely no noise or redundant dimensions. If all 1,000 features of a dataset are equally vital for minimizing the loss, aggressively forcing the smallest weights to absolute zero via truncation will irreparably destroy useful predictive patterns, resulting in substantially higher regret (error) than simply letting SGD keep all weights active but small.
