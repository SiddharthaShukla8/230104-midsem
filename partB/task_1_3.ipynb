# Task 1.3 What the Paper Claims to Improve (9 marks)

*   **Baseline/Prior Method:** Simple Coefficient Rounding (or rounding small coefficients to zero after every $K$ online steps) and the naïve Sub-gradient Algorithm for $L_1$-Regularization.
*   **Limitation Identification:** The paper identifies that the naïve Sub-gradient Algorithm for $L_1$-Regularization fails to produce exact zeroes for floating-point weights, thus providing little practical sparsity. Similarly, Simple Coefficient Rounding is ad-hoc, overly aggressive in discarding weights, very sensitive to the arbitrarily chosen step interval $K$, and fundamentally lacks a robust theoretical guarantee for online regret performance.
*   **How the Proposed Method Attempts to Overcome This Limitation:** The proposed "Truncated Gradient" method bridges the gap between these baseline approaches by introducing a gravity parameter $g$ to provide continuous, gentle shrinkage towards zero (the truncation function $T_1$) along with the stochastic gradient update. This simulates the behavior of $L_1$ regularization efficiently in a mathematically provable online framework, achieving robust sparsity without heavily degrading predictive accuracy.
*   **Condition/Scenario Under Which the Proposed Method Would Not Outperform the Baseline:** The truncated gradient method would likely not outperform exact batch Lasso when applied to a very small, static dataset with dense features. Since the dataset is small enough to fit completely in memory, the exact batch constrained optimization techniques used in Lasso would find the absolute optimal sparse coefficients faster and with a better guarantee of minimizing regularized loss over the whole dataset than an online stochastic approximation.
